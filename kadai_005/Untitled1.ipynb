```python
from os import XATTR_SIZE_MAX
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(2, 3, tight_layout = True)

y0 = [1, 2, -5, 2]

# 0から10までの幅で、配列を等間隔に100個作成する
x1 = np.linspace(0, 10, 100)
y1 = 2 * np.sin(2 * x1)

# 平均が20、標準偏差5の標準正規分布から2000個の乱数を生成
x2 = np.random.normal(20, 5, 2000)

x3 = np.random.normal(0, 20, 60)
y3 = np.random.normal(0, 2, len(x3))

axes[0,0].plot(y0)
axes[0,1].plot(y1)
axes[0,2].hist(x2)
axes[1,0].scatter(x3, y3)

plt.show()
```
